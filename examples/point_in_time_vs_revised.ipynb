# Point-in-time data is not the same as a current historical series

This notebook inspects FinReplay's packaged catalogs without network access. It demonstrates the eligibility distinction; it does not claim forecast skill or investment performance.

```python
from finreplay.catalog import load_adapter_catalog, load_scenario_catalog

adapters = load_adapter_catalog()
scenarios = load_scenario_catalog()
adapters.adapter_count, scenarios.scenario_count
```

A successful live validation proves that a response matched its contract at the recorded retrieval time. Only entries with a separately supported historical boundary can be used as historical decision-time evidence.

```python
from collections import Counter

(
    Counter(a.temporal_coverage for a in adapters.adapters),
    [a.adapter_id for a in adapters.adapters if a.historical_replay_eligible],
)
```

The scenario catalog is a separate surface: each entry points to an exact locked input set and a recorded pack hash. Reproducing one scenario tests repository behavior against that identity; it does not constitute external method review.

```python
svb = next(s for s in scenarios.scenarios if s.slug == "svb-2023")
{
    "decision_time": svb.decision_time,
    "locked_records": svb.distinct_input_records,
    "pack_sha256": svb.pack_sha256,
}
```
